```python
from playwright.async_api import async_playwright
from bs4 import BeautifulSoup
import re
import asyncio

async def fetch_js_page(url):
    async with async_playwright() as p:
        browser = await p.chromium.launch(headless=True)
        page = await browser.new_page()
        await page.goto(url, wait_until="domcontentloaded")
        text = await page.inner_text("body")
        await browser.close()
        response = await clean_extracted_text(text)
        return response

# response = await fetch_js_page(
#     "https://afr100.org/project/regreening-south-africa-sustainable-future"
# )
async def clean_extracted_text(response):
  soup = BeautifulSoup(response, "html.parser")

  # --- 1️⃣ Remove the cookie banner node by CSS selector ---
  # Inspect the page (right-click → Inspect) and look for
  # a container with a name like “cookie”, “consent”, etc.
  for banner in soup.select('[id*="cookie"], [class*="cookie"], [class*="consent"]'):
      banner.decompose()

  # --- 2️⃣ Get the text ---
  text = soup.get_text(separator="\n", strip=True)

  # --- 3️⃣ (Optional) Extra safety: remove any leftover lines
  # that contain typical cookie-banner wording.
  patterns = [
      r"This website stores data such as cookies.*?(Accept|Deny|Manage)",
      r"Privacy Policy",   # if you don’t want this phrase at all
  ]

  for pat in patterns:
      text = re.sub(pat, "", text, flags=re.I | re.S)

  # Collapse multiple blank lines
  text = re.sub(r"\n\s*\n", "\n\n", text).strip()
  return text
  # print("FINAL TEXT", text)




```

```python
from sentence_transformers import SentenceTransformer, util
# model = SentenceTransformer('all-MiniLM-L6-v2')
sentence_transformer
def semantic_score(a, b):
    if not a or not b:
        return 0.0
    emb1 = sentence_transformer.encode(a, convert_to_tensor=True)
    emb2 = sentence_transformer.encode(b, convert_to_tensor=True)
    return float(util.cos_sim(emb1, emb2).diagonal().cpu().numpy())
```

```python
# Install required packages first (if not installed):
# pip install pandas spacy sentence-transformers scikit-learn openai plotly
# python -m spacy download en_core_web_sm

import pandas as pd
import spacy
from sklearn.cluster import KMeans
from sentence_transformers import SentenceTransformer, util
from openai import OpenAI
import numpy as np
import plotly.express as px
import re
from collections import Counter

# 🔑 Initialize
nlp = spacy.load("en_core_web_sm")
# embedder = SentenceTransformer('all-MiniLM-L6-v2')
# client = OpenAI()  # assumes your OPENAI_API_KEY is set in environment
client = ChatOpenAI(model="gpt-5-2025-08-07", temperature=0, openai_api_key=OPENAI_KEY)
# ------------------------------------
# 1️⃣ Load the three model outputs
# ------------------------------------
models = {
    "gpt4o-mini": gpt_data_df,
    "llama3.1": llama_data_df,
    "qwen2.5": qwen_data_df,
}

fields = ["communities", "interventions", "outcomes", "impacts", "time_frames"]

# ------------------------------------
# 2️⃣ Traditional NLP Cleaning
# ------------------------------------
def normalize_text(text):
    if not isinstance(text, str): return ""
    text = text.lower()
    text = re.sub(r'[^a-z\s]', ' ', text)
    return re.sub(r'\s+', ' ', text).strip()

def extract_noun_phrases(text):
    doc = nlp(text)
    return [chunk.text for chunk in doc.noun_chunks if len(chunk.text.split()) >= 2]

# Apply to all models
for model_name, df in models.items():
    for fld in fields:
        col = fld + "_clean"
        df[col] = df[fld].astype(str).apply(normalize_text)
        df[fld + "_phrases"] = df[col].apply(extract_noun_phrases)

# ------------------------------------
# 3️⃣ Build a Combined Phrase Corpus
# ------------------------------------
phrase_records = []
for model_name, df in models.items():
    for fld in fields:
      for _, row in df.iterrows():
        for phrase in row[fld + "_phrases"]:
            phrase_records.append({
                "document_id": row["document_id"],
                "model": model_name,
                "field": fld,
                "phrase": phrase
                })
            # for p in phrases:

phrase_df = pd.DataFrame(phrase_records)
print(f"Total phrases collected: {len(phrase_df)}")

# ------------------------------------
# 4️⃣ Embeddings + Clustering
# ------------------------------------
all_phrases = phrase_df["phrase"].tolist()
embeddings = sentence_transformer.encode(all_phrases, convert_to_tensor=True)
num_clusters = min(10, len(all_phrases)) # adjust depending on variety of data
kmeans = KMeans(n_clusters=num_clusters, random_state=42, n_init=10)
labels = kmeans.fit_predict(embeddings.cpu().numpy())
phrase_df["cluster"] = labels

# ------------------------------------
# 5️⃣ GenAI-as-a-Judge: Cluster Summarization
# ------------------------------------
def summarize_cluster(phrases):
    text_sample = "; ".join(phrases[:10])  # send up to 10 phrases for context
    prompt = f"""
    The following phrases describe extractions from climate-related studies:
    {text_sample}
    Please summarize them into one short representative label that best captures their shared meaning.
    Respond only with the label, nothing else.
    """
    print(f"Phrases: {text_sample}")
    try:
        messages=[{"role": "user", "content": prompt}]
        response = client.invoke(messages)
        return response.content.strip()
    except Exception as e:
        print("Judge error:", e)
        return "unspecified cluster"

cluster_labels = phrase_df.groupby("cluster")["phrase"].apply(list).apply(summarize_cluster)
label_map = cluster_labels.to_dict()
phrase_df["cluster_label"] = phrase_df["cluster"].map(label_map)


# --------------------------------
# 6️⃣ Compute Presence Matrix (Cluster × Field)
# --------------------------------
presence = (
    phrase_df.groupby(["cluster_label", "field"])["document_id"]
    .nunique()
    .reset_index(name="doc_count")
)
presence_matrix = presence.pivot_table(
    index="cluster_label", columns="field", values="doc_count", fill_value=0
)
presence_norm = presence_matrix.div(presence_matrix.sum(axis=1), axis=0)


#OLD
# ------------------------------------
# 6️⃣ Evaluate Semantic Agreement per Model
# ------------------------------------
# summary_records = []
# for fld in fields:
#     for cluster_id, group in phrase_df[phrase_df["field"] == fld].groupby("cluster_label"):
#         avg_sim = 0
#         count = 0
#         for model_name in models.keys():
#             subset = group[group["model"] == model_name]
#             if not subset.empty:
#                 reps = subset["phrase"].tolist()
#                 reps_emb = sentence_transformer.encode(reps, convert_to_tensor=True)
#                 cluster_emb = sentence_transformer.encode(cluster_id, convert_to_tensor=True)
#                 sim = float(util.cos_sim(reps_emb.mean(dim=0), cluster_emb))
#                 avg_sim += sim
#                 count += 1
#         if count > 0:
#             summary_records.append({
#                 "field": fld,
#                 "cluster": cluster_id,
#                 "mean_similarity": round(avg_sim / count, 3),
#                 "models_contributing": count
#             })

# summary_df = pd.DataFrame(summary_records)

# # ------------------------------------
# # 7️⃣ Visualization: Cluster Agreement
# # ------------------------------------
# fig = px.bar(
#     summary_df.sort_values("mean_similarity", ascending=False),
#     x="cluster",
#     y="mean_similarity",
#     color="field",
#     title="Semantic Agreement Between Models per Extracted Concept",
#     labels={"mean_similarity": "Mean Semantic Similarity", "cluster": "Concept (Cluster Label)"}
# )
# fig.update_layout(xaxis_tickangle=-45)
# fig.show()

# # ------------------------------------
# # 8️⃣ Optional: Heatmap by Field × Model
# # ------------------------------------
# import seaborn as sns
# import matplotlib.pyplot as plt

# heatmap_data = phrase_df.groupby(["field", "model"]).size().unstack(fill_value=0)
# sns.heatmap(heatmap_data, annot=True, cmap="YlGnBu")
# plt.title("Number of Extracted Phrases per Field and Model")
# plt.show()

# # ------------------------------------
# # 9️⃣ Save Output
# # ------------------------------------
# phrase_df.to_csv("hybrid_semantic_clusters.csv", index=False)
# summary_df.to_csv("semantic_agreement_summary.csv", index=False)

# print("\n✅ Pipeline complete. Outputs:")
# print("- hybrid_semantic_clusters.csv (phrase-level data)")
# print("- semantic_agreement_summary.csv (aggregated agreement per concept)")
```

```python
# --- Imports ---
import pandas as pd, numpy as np, json, time
import seaborn as sns, matplotlib.pyplot as plt
from openai import OpenAI

client = OpenAI()

# ============================================================
# 1️⃣  Load your extracted data for each model
# ============================================================
files = {
    "gpt4o-mini": "gpt4o_results.json",
    "llama3.1": "llama_results.json",
    "qwen2.5": "qwen_results.json"
}
client = ChatOpenAI(model="gpt-5-2025-08-07", temperature=0, openai_api_key=OPENAI_KEY)
# ------------------------------------
# 1️⃣ Load the three model outputs
# ------------------------------------

models = {
    "gpt-4o-mini": gpt_data_df,
    "llama3.1": llama_data_df,
    "qwen2.5": qwen_data_df,
}

fields = ["communities", "interventions", "outcomes", "impacts", "time_frames"]

dfs = []
for model, df in models.items():
    # try:
    #     df = pd.read_json(file)
    # except Exception:
    #     df = pd.read_csv(file)
    df["model"] = model
    dfs.append(df)

all_df = pd.concat(dfs, ignore_index=True)

# Ensure expected columns exist
for c in ["communities","interventions","outcomes","impacts"]:
    if c not in all_df.columns: all_df[c] = ""

# ============================================================
# 2️⃣  Define Judge Prompt
# ============================================================
def build_prompt(row):
  print(row.get('communities','') == 'NOT RELATED TO CLIMATE CHANGE')
  if row.get('communities','') == 'NOT RELATED TO CLIMATE CHANGE':
    return
  return f"""
          You are an expert reviewer evaluating climate literacy interventions for SDG 13.3.

          Given the extracted evidence below, rate each dimension 0.0–1.0 and compute the overall Climate Literacy Impact Index (CLII)
          using weights Cognitive 25%, Affective 20%, Behavioral 30%, Institutional 25%.
          Return valid JSON only.

          EVIDENCE
          Community: {row.get('communities','')}
          Intervention: {row.get('interventions','')}
          Outcome: {row.get('outcomes','')}
          Impact: {row.get('impacts','')}
          """
# Time-frame: {row.get('timeframe','')}

# ============================================================
# 3️⃣  LLM-as-Judge scoring
# ============================================================
def judge_clii(row, retries=2):
    prompt = build_prompt(row)
    for _ in tqdm(range(retries)):
        try:
            messages=[{"role": "user", "content": prompt}]
            response = client.invoke(messages)
            result = json.loads(response.content)
            return result
        except Exception as e:
            print("Judge error:", e)
            time.sleep(2)
            last_err = str(e)
            return {"error": last_err, "raw": result if 'content' in locals() else None}

# ============================================================
# 4️⃣  Run judge on sample or full dataset
# ============================================================
# ⚠️ Start with a subset first to control cost, then expand.
subset = all_df#.sample(n=3, random_state=42)
results = subset.apply(judge_clii, axis=1)

# Flatten JSON into columns
scored = pd.concat([subset.reset_index(drop=True),
                    pd.json_normalize(results)], axis=1)

# ============================================================
# 5️⃣  Compute CLII if Judge didn't return it
# ============================================================
weights = {"cognitive":0.25, "affective":0.20, "behavioral":0.30, "institutional":0.25}
def compute_clii(row):
    vals = [row.get(k, np.nan) for k in weights]
    if any(np.isnan(v) for v in vals): return np.nan
    score = sum(weights[k]*row[k] for k in weights)
    return round(100*score/sum(weights.values()),2)

if "clii" not in scored.columns:
    scored["clii"] = scored.apply(compute_clii, axis=1)

scored.to_csv("clii_scored_results.csv", index=False)
print("✅ CLII scoring complete and saved → clii_scored_results.csv")

# ============================================================
# 6️⃣  Aggregate per model for comparison
# ============================================================
agg = scored.groupby("model")[["cognitive","affective","behavioral","institutional","clii"]].mean().round(2)
print(agg)

# ============================================================
# 7️⃣  Visualizations
# ============================================================

# --- Heatmap of average component scores ---
plt.figure(figsize=(8,4))
sns.heatmap(agg[["cognitive","affective","behavioral","institutional"]],
            annot=True, cmap="YlGnBu", fmt=".2f")
plt.title("Average CLII Component Scores per Model")
plt.tight_layout()
plt.show()

# --- Barplot of overall CLII ---
plt.figure(figsize=(6,4))
sns.barplot(x=agg.index, y=agg["clii"], palette="viridis")
plt.ylabel("Mean CLII (0–100)")
plt.title("Overall Climate Literacy Impact Index per Model")
plt.tight_layout()
plt.show()

# --- Radar for one model (e.g., GPT4o-mini) ---
def radar_plot(model_name):
    import numpy as np
    row = agg.loc[model_name]
    cats = ["cognitive","affective","behavioral","institutional"]
    vals = [row[c] for c in cats]; vals += vals[:1]
    angs = np.linspace(0, 2*np.pi, len(cats), endpoint=False).tolist(); angs += angs[:1]
    fig, ax = plt.subplots(figsize=(6,6), subplot_kw=dict(polar=True))
    ax.plot(angs, vals, 'o-', linewidth=2)
    ax.fill(angs, vals, alpha=0.25)
    ax.set_thetagrids(np.degrees(angs[:-1]), [c.capitalize() for c in cats])
    ax.set_ylim(0,1)
    ax.set_title(f"CLII Component Profile – {model_name}")
    plt.show()

radar_plot("gpt-4o-mini")
```

```python
# Required packages:
# pip install pandas scikit-learn matplotlib plotly wordcloud
# Optional (for lemmatization): pip install spacy && python -m spacy download en_core_web_sm

import json
import re
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.feature_extraction import _stop_words
from sklearn.feature_extraction.text import CountVectorizer
from collections import Counter
import plotly.graph_objects as go

# ---------- Config ----------
INPUT_PATH = "llama_user_2_matches.json"   # adjust as needed
FIELDS = ['intervention_llama', 'outcome_llama', 'impact_llama', 'time_frame_llama']
TOP_N = 15                # top terms/phrases to show
MIN_DF = 1                # min doc frequency for vectorizer
SANKey_MIN_LINK = 1       # minimal aggregated link weight to include
sklearn_stop = _stop_words.ENGLISH_STOP_WORDS

# Use sklearn's english stop words plus domain stopwords
domain_stopwords = {
    "project","programme","program","study","report","intervention","interventions",
    "community","communities","participants","people","schools","school","workshop"
}

# Convert to list (important!)
# stop_words = list(set(sklearn_stop) | domain_stopwords)


# Use sklearn's english stop words plus domain stopwords
# from sklearn.feature_extraction import _stop_words
sklearn_stop = _stop_words.ENGLISH_STOP_WORDS
stop_words = set(sklearn_stop) | domain_stopwords

# Optional: use a small function to normalize text
def normalize_text(text):
    if pd.isna(text):
        return ""
    text = str(text).lower()
    # remove URLs, emails
    text = re.sub(r'http\S+|www\.\S+|\S+@\S+', ' ', text)
    # remove punctuation except hyphens (optional)
    text = re.sub(r'[^\w\s-]', ' ', text)
    # collapse whitespace
    text = re.sub(r'\s+', ' ', text).strip()
    return text

# Optional lemmatization (comment out if not installed)
# import spacy
# nlp = spacy.load("en_core_web_sm")
# def lemmatize_text(text):
#     doc = nlp(text)
#     return " ".join([tok.lemma_ for tok in doc if not tok.is_stop and not tok.is_punct])

# ---------- Load data ----------
with open(INPUT_PATH, 'r', encoding='utf-8') as f:
    raw = json.load(f)

df = pd.json_normalize(raw)

# Ensure the fields exist
for fld in FIELDS:
    if fld not in df.columns:
        df[fld] = np.nan

# Normalize fields into cleaned text columns
for fld in FIELDS:
    clean_col = fld + "_clean"
    df[clean_col] = df[fld].astype(str).apply(normalize_text)

# ---------- Helper: get top ngrams ----------
def top_ngrams(corpus_series, n=TOP_N, ngram_range=(1,1), stop_words=stop_words, min_df=MIN_DF):
    """
    Returns top n ngrams and counts for a pandas Series (corpus of texts).
    Handles empty/stopword-only documents gracefully.
    """
    docs = corpus_series.dropna().astype(str)
    # Clean up whitespace and ignore empty docs
    docs = docs[docs.str.strip() != ""]
    if len(docs) == 0:
        print(f"⚠️ Skipping n-gram extraction: no valid documents in corpus (range={ngram_range})")
        return []
    try:
        vect = CountVectorizer(
            ngram_range=ngram_range,
            stop_words=list(stop_words),
            min_df=min_df,
            token_pattern=r'\b[a-zA-Z]{3,}\b'  # only keep alphabetic words with >=3 letters
        )
        X = vect.fit_transform(docs)
        if X.shape[1] == 0:
            print(f"⚠️ No valid vocabulary (maybe all stop words?) — skipping.")
            return []
        sums = np.asarray(X.sum(axis=0)).ravel()
        terms = vect.get_feature_names_out()
        counts = list(zip(terms, sums))
        counts_sorted = sorted(counts, key=lambda x: x[1], reverse=True)
        return counts_sorted[:n]
    except ValueError as e:
        print(f"⚠️ Skipping corpus due to error: {e}")
        return []

# ---------- Compute top unigrams & bigrams per field ----------
results = {}
for fld in FIELDS:
    corpus = df[fld + "_clean"]
    top_unigrams = top_ngrams(corpus, n=TOP_N, ngram_range=(1,1))
    top_bigrams  = top_ngrams(corpus, n=TOP_N, ngram_range=(2,2))
    results[fld] = {"unigrams": top_unigrams, "bigrams": top_bigrams}

# Save top terms to CSV for inspection
out_top_terms = []
for fld, vals in results.items():
    for term, cnt in vals["unigrams"]:
        out_top_terms.append({"field": fld, "ngram": term, "count": int(cnt), "type": "unigram"})
    for term, cnt in vals["bigrams"]:
        out_top_terms.append({"field": fld, "ngram": term, "count": int(cnt), "type": "bigram"})
pd.DataFrame(out_top_terms).to_csv("top_terms_per_field.csv", index=False)

# ---------- Plot horizontal bar charts (top unigrams) ----------
for fld in FIELDS:
    data = results[fld]["unigrams"]
    if not data:
        continue
    terms, counts = zip(*data)
    plt.figure(figsize=(8, max(3, len(terms)*0.35)))
    plt.barh(terms[::-1], counts[::-1], color="steelblue")
    plt.title(f"Top {len(terms)} words in {fld}")
    plt.xlabel("Frequency")
    plt.tight_layout()
    plt.show()

# ---------- Build Sankey from aggregated Intervention->Outcome->Impact connections ----------
# We'll aggregate by string exact-match; you can improve by normalizing terms or clustering (fuzzy match)
sources, targets, values = [], [], []

for _, row in df.iterrows():
    inter = row.get("intervention_llama")
    out = row.get("outcome_llama")
    imp = row.get("impact_llama")
    # prefer cleaned shorter versions if available
    if pd.notna(inter) and str(inter).strip():
        s = str(inter).strip()
    else:
        s = None
    if pd.notna(out) and str(out).strip():
        t = str(out).strip()
    else:
        t = None
    if pd.notna(imp) and str(imp).strip():
        u = str(imp).strip()
    else:
        u = None
    if s and t:
        sources.append((s,t))
    if t and u:
        sources.append((t,u))

# Aggregate counts
link_counts = Counter(sources)
# Filter small links to reduce clutter
link_counts = {k:v for k,v in link_counts.items() if v >= SANKey_MIN_LINK}

if link_counts:
    # labels (unique nodes)
    nodes = list({n for pair in link_counts.keys() for n in pair})
    label_index = {label:i for i,label in enumerate(nodes)}
    src_ids = [label_index[pair[0]] for pair in link_counts.keys()]
    tgt_ids = [label_index[pair[1]] for pair in link_counts.keys()]
    vals = [v for v in link_counts.values()]

    sankey_fig = go.Figure(data=[go.Sankey(
        node = dict(label=nodes, pad=15, thickness=18),
        link = dict(source=src_ids, target=tgt_ids, value=vals)
    )])
    sankey_fig.update_layout(title_text="Intervention → Outcome → Impact (aggregated)", font_size=10)
    sankey_fig.write_html("sankey_intervention_outcome_impact.html")
    print("Sankey saved to sankey_intervention_outcome_impact.html")
else:
    print("No Sankey links found (maybe few non-empty intervention/outcome/impact triples).")

# ---------- Optional: small network visualization (matplotlib + networkx) ----------
try:
    import networkx as nx
    G = nx.DiGraph()
    for (a, b), cnt in link_counts.items():
        G.add_edge(a, b, weight=cnt)
    plt.figure(figsize=(10,8))
    pos = nx.spring_layout(G, k=0.5, iterations=50)
    edge_weights = [G[u][v]['weight'] for u,v in G.edges()]
    nx.draw(G, pos, with_labels=True, node_size=800, font_size=9, arrowsize=12)
    nx.draw_networkx_edge_labels(G, pos, edge_labels=nx.get_edge_attributes(G,'weight'), font_color='gray', font_size=8)
    plt.title("Intervention→Outcome→Impact network (aggregated)")
    plt.tight_layout()
    plt.show()
except Exception as e:
    print("networkx not available or failed; skipping network plot.", str(e))

# ---------- Print short summary for researcher review ----------
print("\n=== Summary ===")
for fld in FIELDS:
    non_null = df[fld].notna().sum()
    print(f"{fld}: {non_null} non-empty entries")
print("\nTop terms saved to top_terms_per_field.csv")
print("Sankey (if built) saved to sankey_intervention_outcome_impact.html")

# Save cleaned dataframe as CSV for manual inspection
df.to_csv("llama_user_2_matches_cleaned.csv", index=False)
```

```python
import tiktoken

def count_tokens(text, model="gpt-4o-mini"):
    """Count tokens in a string for a given OpenAI model."""
    encoding = tiktoken.encoding_for_model(model)
    return len(encoding.encode(text))

def average_token_count(prompts, model="gpt-4o-mini"):
    """Get average token count from a list of prompts."""
    total_tokens = sum(count_tokens(prompt['context'] +" "+ prompt['question'] +" "+ prompt['answer'], model) for prompt in prompts)
    avg_tokens = total_tokens / len(prompts) if prompts else 0
    return {
        "average_tokens": avg_tokens,
        "total_tokens": total_tokens,
        "prompt_count": len(prompts)
    }

# result = average_token_count(outputs, model="gpt-4o-mini")
result = average_token_count(eval_interv_4_gpt_data, model="gpt-4o-mini")
print(f"Average tokens per prompt: {result['average_tokens']:.2f}")
print(f"Total tokens: {result['total_tokens']}")
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import precision_score, recall_score

def compute_precision_recall(context, answer):
    vectorizer = TfidfVectorizer().fit([context, answer])
    context_vec = vectorizer.transform([context]).toarray()[0]
    answer_vec = vectorizer.transform([answer]).toarray()[0]

    context_words = set(i for i, val in enumerate(context_vec) if val > 0)
    answer_words = set(i for i, val in enumerate(answer_vec) if val > 0)

    if not answer_words:
        return 0.0, 0.0

    tp = len(context_words & answer_words)
    precision = tp / len(answer_words)
    recall = tp / len(context_words) if context_words else 0.0

    return round(precision, 4), round(recall, 4)
```

```python
from typing import Any, Dict, List

from langchain.callbacks.manager import CallbackManagerForLLMRun
from langchain.schema import Generation, LLMResult
from langchain.llms.base import BaseLLM

class RunnableLLMWrapper(BaseLLM):

    def __init__(self, llm):
        super().__init__()
        self.llm = llm

    @property
    def _llm(self) -> Dict[str, Any]:
        return self.llm

    @property
    def _identifying_params(self) -> Dict[str, Any]:
        return {}

    @property
    def _llm_type(self) -> str:
        return "runnable_llm_wrapper"

    def _call(
        self,
        prompt: str,
        stop: Optional[List[str]] = None,
        run_manager: Optional[CallbackManagerForLLMRun] = None,
        **kwargs: Any,
    ) -> str:
        response = self.llm(prompt, **kwargs)
        return response

    async def _acall(
            self,
            prompt: str,
            stop: Optional[List[str]] = None,
            run_manager: Optional[CallbackManagerForLLMRun] = None,
            **kwargs: Any,
        ) -> str:
            """Call out to an underlying LLM."""
            # response = await self.llm.acall(prompt, **kwargs)
            response = self.llm.invoke(prompt, **kwargs)
            return response

    def _generate(
        self,
        prompts: List[str],
        stop: Optional[List[str]] = None,
        run_manager: Optional[CallbackManagerForLLMRun] = None,
        **kwargs: Any,
    ) -> LLMResult:
        """Generate text from prompts."""
        # Implement your logic here to generate text using self.llm
        # This is a placeholder, replace it with your actual implementation
        generations = []
        for prompt in prompts:
            response = self._call(prompt, stop=stop, run_manager=run_manager, **kwargs)
            generations.append([Generation(text=response)])
        return LLMResult(generations=generations)
```

```python
from langchain.evaluation import load_evaluator
from langchain.embeddings import HuggingFaceEmbeddings
# from langchain.llms import HuggingFaceHub, ChatOpenAI
from langchain_openai import ChatOpenAI
from ragas.llms import LangchainLLMWrapper
from langchain.evaluation.schema import LLMEvalChain
import numpy as np
import pandas as pd

# === Embedding model for semantic similarity
# embedding_model = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
embedding_model = embeddings #SentenceTransformer('thenlper/gte-small')

deepseek_llm = HuggingFaceEndpoint(
    repo_id="deepseek-ai/DeepSeek-R1-Distill-Qwen-32B",
    task="text-generation",
    temperature= 0.2,
    max_new_tokens= 512,
    huggingfacehub_api_token=HF_KEY,

    )


llm_wrapper = RunnableLLMWrapper(deepseek_llm)

# === Evaluation Chains (can be customized per model)
faithfulness_chain = load_evaluator("context_qa", llm=llm_wrapper)
relevance_chain = load_evaluator("qa", llm=llm_wrapper)
# context_precision_chain = load_evaluator("context_precision", llm=llm_wrapper)
# context_recall_chain = load_evaluator("context_recall", llm=llm_wrapper)

# === Semantic similarity
def semantic_similarity(text1, text2):
    emb1 = embedding_model.embed_query(text1)
    emb2 = embedding_model.embed_query(text2)
    return np.dot(emb1, emb2) / (np.linalg.norm(emb1) * np.linalg.norm(emb2))

# === Evaluation Wrapper Function
def evaluate_response(question, context, reference_answer, model_answer, verbose=False):
    # Setup example + prediction format
    example = {
        "query": question,
        "context": context,
        "answer": reference_answer,
    }
    prediction = {
        "result": model_answer
    }

    results = {}

    # Faithfulness
    faithful = faithfulness_chain.evaluate([example], [prediction])[0]
    results["faithfulness"] = faithful

    # Answer Relevance
    relevance = relevance_chain.evaluate([example], [prediction])[0]
    results["answer_relevance"] = relevance

    # # Context Precision
    # context_prec = context_precision_chain.evaluate([example], [prediction])[0]
    # results["context_precision"] = context_prec

    # # Context Recall
    # context_rec = context_recall_chain.evaluate([example], [prediction])[0]
    # results["context_recall"] = context_rec

    # Semantic Similarity
    sim = semantic_similarity(reference_answer, model_answer)
    results["semantic_similarity"] = round(sim, 4)

    if verbose:
        print(pd.DataFrame([results]))

    return results
```
